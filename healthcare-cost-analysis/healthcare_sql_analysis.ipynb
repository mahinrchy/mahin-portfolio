```python
import pandas as pd

url = "https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/insurance.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
import sqlite3

conn = sqlite3.connect("healthcare.db")
df.to_sql("insurance", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT region, ROUND(AVG(charges),2) AS avg_cost
FROM insurance
GROUP BY region
ORDER BY avg_cost DESC;
"""

import pandas as pd
pd.read_sql(query, conn)
```

```python
query = """
SELECT smoker, ROUND(AVG(charges),2) AS avg_cost
FROM insurance
GROUP BY smoker;
"""

pd.read_sql(query, conn)
```

Key Insights

1. The Southeast region has the highest average medical charges among all regions.
2. Smoking status is the strongest cost driver in the dataset.
3. Smokers incur average healthcare costs of 32,050.23, compared with 8,434.27 for non-smokers.
4. On average, smokers cost nearly four times more than non-smokers.

```python
query = """
SELECT
  CASE
    WHEN bmi < 18.5 THEN 'Underweight'
    WHEN bmi BETWEEN 18.5 AND 24.9 THEN 'Normal'
    WHEN bmi BETWEEN 25 AND 29.9 THEN 'Overweight'
    ELSE 'Obese'
  END AS bmi_category,
  ROUND(AVG(charges),2) AS avg_cost
FROM insurance
GROUP BY bmi_category
ORDER BY avg_cost DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
  CASE
    WHEN age < 30 THEN 'Young'
    WHEN age BETWEEN 30 AND 50 THEN 'Middle Age'
    ELSE 'Senior'
  END AS age_group,
  ROUND(AVG(charges),2) AS avg_cost
FROM insurance
GROUP BY age_group
ORDER BY avg_cost DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT age, sex, bmi, children, smoker, region, charges
FROM insurance
ORDER BY charges DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

Final Insights

1. Smoking is the strongest predictor of high healthcare costs, with smokers incurring nearly four times higher average charges than non-smokers.

2. Obesity significantly increases healthcare expenses, with obese individuals having the highest average cost among all BMI categories.

3. Healthcare costs increase consistently with age, with senior individuals incurring the highest expenses.

4. The most expensive patients are predominantly smokers with high BMI, indicating a compounding effect of risk factors.

5. Regional differences exist, with the Southeast showing the highest average costs, although the variation is less significant compared to smoking and BMI.

Business Recommendations

1. Invest in smoking cessation programs to reduce high-cost patient segments.

2. Promote preventive health initiatives targeting obesity and high BMI individuals.

3. Focus early intervention strategies on middle-aged customers before they transition into higher-cost senior groups.

4. Develop targeted healthcare plans for high-risk segments combining age, smoking status, and BMI.

5. Allocate healthcare resources more efficiently in high-cost regions such as the Southeast.
