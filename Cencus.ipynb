```python
import os

folder = r"C:\Users\Mahin\OneDrive\Desktop\Group 18-Phase 2"

print(os.listdir(folder))

df_311_data = pd.read_csv(os.path.join(folder, "311_data.csv"))
df_income_data = pd.read_csv(os.path.join(folder, "income_data.csv"))
df_zip_data = pd.read_csv(os.path.join(folder, "zip_shape.csv"))

print("311 data:")
display(df_311_data.head())

print("Income data:")
display(df_income_data.head())

print("ZIP shape data:")
display(df_zip_data.head())
```

```python
# Keep only useful columns using your actual column names
df_311 = df_311_data[[
    'Created Date',
    'Closed Date',
    'Incident Zip',
    'Problem (formerly Complaint Type)',
    'Borough'
]].copy()

# Rename columns to simpler names
df_311 = df_311.rename(columns={
    'Created Date': 'created_date',
    'Closed Date': 'closed_date',
    'Incident Zip': 'zip_code',
    'Problem (formerly Complaint Type)': 'complaint_type',
    'Borough': 'borough'
})

# Convert dates
df_311['created_date'] = pd.to_datetime(df_311['created_date'], errors='coerce')
df_311['closed_date'] = pd.to_datetime(df_311['closed_date'], errors='coerce')

# Remove missing important values
df_311 = df_311.dropna(subset=['created_date', 'closed_date', 'zip_code'])

# Create resolution time in hours
df_311['resolution_time_hours'] = (
    df_311['closed_date'] - df_311['created_date']
).dt.total_seconds() / 3600

# Remove bad values
df_311 = df_311[
    (df_311['resolution_time_hours'] > 0) &
    (df_311['resolution_time_hours'] < 500)
]

# Clean ZIP code
df_311['zip_code'] = df_311['zip_code'].astype(str).str[:5]

# Check result
display(df_311.head())
print(df_311.shape)
```

```python
# Reload income data
df_income_data = pd.read_csv("income_data.csv")

# Show columns
print(df_income_data.columns.tolist())

# Find ZIP column
zip_col = "NAME"

# Find the median household income column automatically
income_cols = [col for col in df_income_data.columns if "S1903_C03_001E" in col or "Median income" in col]

print("Possible income columns:", income_cols)
```

```python
df_income = df_income_data[["NAME", "S1903_C03_001E"]].copy()

df_income.columns = ["zip_code", "median_income"]

df_income["zip_code"] = df_income["zip_code"].astype(str).str.replace("ZCTA5 ", "", regex=False)
df_income["median_income"] = pd.to_numeric(df_income["median_income"], errors="coerce")

df_income = df_income.dropna()

print(df_income.shape)
display(df_income.head())
```

```python
df_income = df_income[df_income["zip_code"] != "New York"]
```

```python
df_merged = df_311.merge(df_income, on="zip_code")

print(df_merged.shape)
display(df_merged.head())
```

```python
df_merged["income_group"] = pd.qcut(
    df_merged["median_income"],
    q=3,
    labels=["Low Income", "Middle Income", "High Income"]
)

display(df_merged.head())
print(df_merged["income_group"].value_counts())
```

```python
import plotly.express as px

df_scatter = df_merged.groupby("zip_code").agg({
    "median_income": "mean",
    "resolution_time_hours": "mean"
}).reset_index()

fig2 = px.scatter(
    df_scatter,
    x="median_income",
    y="resolution_time_hours",
    trendline="ols",
    title="Median Income vs Average Resolution Time",
    labels={
        "median_income": "Median Household Income ($)",
        "resolution_time_hours": "Avg Resolution Time (Hours)"
    }
)

fig2.show()
```

This scatterplot shows the relationship between median income and average 311 resolution time at the ZIP code level. The trendline slopes slightly downward, suggesting that higher-income neighborhoods may receive somewhat faster service. However, the relationship is relatively weak, indicating that income alone does not strongly determine service speed.

```python
# Create month column
df_merged["month"] = df_merged["created_date"].dt.to_period("M").astype(str)

# Aggregate by month + income group
df_time = df_merged.groupby(["month", "income_group"]).agg({
    "resolution_time_hours": "mean"
}).reset_index()

# Plot
fig3 = px.line(
    df_time,
    x="month",
    y="resolution_time_hours",
    color="income_group",
    title="311 Resolution Time Trends Over Time by Income Group",
    labels={
        "resolution_time_hours": "Avg Resolution Time (Hours)",
        "month": "Month",
        "income_group": "Income Group"
    }
)

fig3.show()
```

This chart shows how resolution times change over time across income groups. Low-income neighborhoods consistently experience longer resolution times, while high-income areas tend to receive faster service. This suggests a persistent service gap over time.

```python
import pandas as pd
import plotly.express as px

top_complaints = df_merged["complaint_type"].value_counts().head(5).index

df_bar = df_merged[df_merged["complaint_type"].isin(top_complaints)].copy()

df_bar = df_bar.groupby(["complaint_type", "income_group"]).agg({
    "resolution_time_hours": "mean"
}).reset_index()

df_bar["complaint_type"] = df_bar["complaint_type"].str.title()

# FIX: enforce correct income order
df_bar["income_group"] = pd.Categorical(
    df_bar["income_group"],
    categories=["Low Income", "Middle Income", "High Income"],
    ordered=True
)

# SORT properly
df_bar = df_bar.sort_values(["complaint_type", "income_group"])

fig4 = px.bar(
    df_bar,
    x="complaint_type",
    y="resolution_time_hours",
    color="income_group",
    barmode="group",
    title="Average 311 Resolution Time by Complaint Type and Income Group",
    labels={
        "resolution_time_hours": "Avg Resolution Time (Hours)",
        "complaint_type": "Complaint Type",
        "income_group": "Income Group"
    },
    color_discrete_map={
        "Low Income": "#d62728",
        "Middle Income": "#ff7f0e",
        "High Income": "#2ca02c"
    },
    category_orders={
        "income_group": ["Low Income", "Middle Income", "High Income"]
    }
)

fig4.update_traces(
    texttemplate="%{y:.1f}",
    textposition="outside"
)

fig4.update_layout(
    xaxis_tickangle=-45,
    height=550,
    margin=dict(l=40, r=40, t=70, b=90),
    legend_title="Income Group"
)

fig4.show()
```

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

geojson_url = "https://raw.githubusercontent.com/OpenDataDE/State-zip-code-GeoJSON/master/ny_new_york_zip_codes_geo.min.json"

app = Dash(__name__)

borough_options = sorted(df_merged["borough"].dropna().unique())
income_options = ["Low Income", "Middle Income", "High Income"]
complaint_options = df_merged["complaint_type"].value_counts().head(5).index.tolist()

def filter_data(selected_boroughs, selected_complaints, selected_income, start_date, end_date):
    dff = df_merged.copy()

    dff = dff[
        (dff["created_date"] >= pd.to_datetime(start_date)) &
        (dff["created_date"] <= pd.to_datetime(end_date))
    ]

    if selected_boroughs:
        dff = dff[dff["borough"].isin(selected_boroughs)]

    if selected_complaints:
        dff = dff[dff["complaint_type"].isin(selected_complaints)]

    if selected_income:
        dff = dff[dff["income_group"].isin(selected_income)]

    return dff


def make_map(dff):
    all_zip = df_merged.groupby("zip_code").agg({
        "resolution_time_hours": "mean"
    }).reset_index()

    selected_zip = dff.groupby("zip_code").agg({
        "resolution_time_hours": "mean",
        "median_income": "mean"
    }).reset_index()

    base = go.Choropleth(
        geojson=geojson_url,
        locations=all_zip["zip_code"],
        z=[1] * len(all_zip),
        featureidkey="properties.ZCTA5CE10",
        colorscale=[[0, "#d9d9d9"], [1, "#d9d9d9"]],
        showscale=False,
        marker_line_color="white",
        marker_line_width=0.4,
        hoverinfo="skip"
    )

    color = go.Choropleth(
        geojson=geojson_url,
        locations=selected_zip["zip_code"],
        z=selected_zip["resolution_time_hours"],
        featureidkey="properties.ZCTA5CE10",
        colorscale="RdYlGn_r",
        marker_line_color="white",
        marker_line_width=0.5,
        colorbar=dict(
            title="Avg Resolution<br>Time (Hours)",
            thickness=12,
            len=0.55,
            x=0.92,
            y=0.50
        ),
        hovertemplate=
            "ZIP: %{location}<br>" +
            "Avg Resolution Time: %{z:.1f} hours<br>" +
            "<extra></extra>"
    )

    fig_map = go.Figure(data=[base, color])

    fig_map.update_geos(
        fitbounds="locations",
        visible=False
    )

    fig_map.update_layout(
        title="Average 311 Resolution Time by ZIP Code",
        height=330,
        margin=dict(l=5, r=5, t=35, b=5)
    )

    return fig_map


def make_scatter(dff):
    df_scatter = dff.groupby(["zip_code", "borough"]).agg({
        "median_income": "mean",
        "resolution_time_hours": "mean"
    }).reset_index()

    fig_scatter = px.scatter(
        df_scatter,
        x="median_income",
        y="resolution_time_hours",
        color="borough",
        trendline="ols",
        title="Median Income vs Average Resolution Time",
        labels={
            "median_income": "Median Household Income ($)",
            "resolution_time_hours": "Avg Resolution Time (Hours)",
            "borough": "Borough"
        }
    )

    fig_scatter.update_layout(height=330, margin=dict(l=40, r=10, t=35, b=40))
    return fig_scatter


def make_time(dff):
    dff = dff.copy()
    dff["month"] = dff["created_date"].dt.to_period("M").astype(str)

    df_time = dff.groupby(["month", "income_group"]).agg({
        "resolution_time_hours": "mean"
    }).reset_index()

    fig_time = px.line(
        df_time,
        x="month",
        y="resolution_time_hours",
        color="income_group",
        title="311 Resolution Time Trends Over Time by Income Group",
        labels={
            "resolution_time_hours": "Avg Resolution Time (Hours)",
            "month": "Month",
            "income_group": "Income Group"
        },
        color_discrete_map={
            "Low Income": "#d62728",
            "Middle Income": "#ff7f0e",
            "High Income": "#2ca02c"
        }
    )

    fig_time.update_layout(height=230, margin=dict(l=40, r=10, t=35, b=35))
    return fig_time


def make_bar(dff):
    df_bar = dff[dff["complaint_type"].isin(complaint_options)].copy()

    df_bar = df_bar.groupby(["complaint_type", "income_group"]).agg({
        "resolution_time_hours": "mean"
    }).reset_index()

    df_bar["complaint_type"] = df_bar["complaint_type"].str.title()

    fig_bar = px.bar(
        df_bar,
        x="complaint_type",
        y="resolution_time_hours",
        color="income_group",
        barmode="group",
        title="Average Resolution Time by Complaint Type and Income Group",
        labels={
            "resolution_time_hours": "Avg Resolution Time (Hours)",
            "complaint_type": "Complaint Type",
            "income_group": "Income Group"
        },
        color_discrete_map={
            "Low Income": "#d62728",
            "Middle Income": "#ff7f0e",
            "High Income": "#2ca02c"
        },
        category_orders={
            "income_group": ["Low Income", "Middle Income", "High Income"]
        }
    )

    fig_bar.update_traces(texttemplate="%{y:.1f}", textposition="outside")
    fig_bar.update_layout(
        xaxis_tickangle=-45,
        height=260,
        margin=dict(l=40, r=10, t=35, b=70),
        legend_title="Income Group"
    )

    return fig_bar


card_style = {
    "border": "1px solid #bfc7d1",
    "backgroundColor": "white",
    "padding": "6px",
    "boxSizing": "border-box"
}

header_style = {
    "backgroundColor": "#06264a",
    "color": "white",
    "textAlign": "center",
    "fontWeight": "bold",
    "padding": "5px",
    "fontSize": "14px"
}

insight_bottom_style = {
    "fontSize": "12px",
    "lineHeight": "1.35",
    "padding": "8px",
    "backgroundColor": "white",
    "borderTop": "1px solid #bfc7d1"
}

app.layout = html.Div([

    html.Div([
        html.H1("NYC 311 SERVICE EQUITY DASHBOARD", style={"margin": "0", "fontSize": "28px"}),
        html.Div(
            "Do wealthier neighborhoods receive faster resolution times for similar 311 complaints in New York City?",
            style={"fontStyle": "italic", "fontSize": "15px", "marginTop": "4px"}
        )
    ], style={
        "backgroundColor": "#06264a",
        "color": "white",
        "textAlign": "center",
        "padding": "15px",
        "fontFamily": "Arial"
    }),

    html.Div([

        html.Div([
            html.H3("FILTERS", style={"textAlign": "center", "fontSize": "16px"}),

            html.Label("Borough", style={"fontWeight": "bold"}),
            dcc.Checklist(
                id="borough_filter",
                options=[{"label": b.title(), "value": b} for b in borough_options],
                value=borough_options,
                style={"fontSize": "11px", "marginTop": "8px", "marginBottom": "18px"},
                inputStyle={"marginRight": "5px"}
            ),

            html.Label("Complaint Type", style={"fontWeight": "bold"}),
            dcc.Checklist(
                id="complaint_filter",
                options=[{"label": c.title(), "value": c} for c in complaint_options],
                value=complaint_options,
                style={"fontSize": "11px", "marginTop": "8px", "marginBottom": "18px"},
                inputStyle={"marginRight": "5px"}
            ),

            html.Label("Date Range", style={"fontWeight": "bold"}),
            dcc.DatePickerRange(
                id="date_filter",
                start_date=df_merged["created_date"].min(),
                end_date=df_merged["created_date"].max(),
                display_format="M/D/YYYY",
                style={"fontSize": "11px", "marginBottom": "18px"}
            ),

            html.Br(),
            html.Br(),

            html.Label("Income Group", style={"fontWeight": "bold"}),
            dcc.Checklist(
                id="income_filter",
                options=[{"label": i, "value": i} for i in income_options],
                value=income_options,
                style={"fontSize": "11px", "marginTop": "8px"},
                inputStyle={"marginRight": "5px"}
            )

        ], style={
            "width": "190px",
            "minWidth": "190px",
            "backgroundColor": "#f2f4f7",
            "padding": "12px",
            "borderRight": "1px solid #bfc7d1",
            "fontFamily": "Arial",
            "fontSize": "12px",
            "minHeight": "900px"
        }),

        html.Div([

            html.Div([

                html.Div([
                    html.Div("1. AVERAGE 311 RESOLUTION TIME BY ZIP CODE", style=header_style),
                    dcc.Graph(id="map_chart", config={"displayModeBar": False}),
                    html.Div([
                        html.B("What it shows: "),
                        html.Span("Average resolution time across NYC ZIP codes. "),
                        html.Br(),
                        html.B("Purpose: "),
                        html.Span("Identify where services are faster or slower geographically. "),
                        html.Br(),
                        html.B("Example insight: "),
                        html.Span("Some ZIP codes have much longer resolution times than others.")
                    ], style=insight_bottom_style)
                ], style={**card_style, "width": "50%"}),

                html.Div([
                    html.Div("2. MEDIAN INCOME VS AVERAGE RESOLUTION TIME", style=header_style),
                    dcc.Graph(id="scatter_chart", config={"displayModeBar": False}),
                    html.Div([
                        html.B("What it shows: "),
                        html.Span("Relationship between income and resolution time by ZIP code. "),
                        html.Br(),
                        html.B("Purpose: "),
                        html.Span("Determine whether higher-income areas receive faster service. "),
                        html.Br(),
                        html.B("Example insight: "),
                        html.Span("The downward trend suggests wealthier areas may receive somewhat faster service.")
                    ], style=insight_bottom_style)
                ], style={**card_style, "width": "50%"})

            ], style={"display": "flex", "gap": "8px", "marginBottom": "8px"}),

            html.Div([
                html.Div("3. 311 RESOLUTION TIME TRENDS OVER TIME BY INCOME GROUP", style=header_style),
                dcc.Graph(id="time_chart", config={"displayModeBar": False}),
                html.Div([
                    html.B("What it shows: "),
                    html.Span("Change in resolution time over time by income group. "),
                    html.Br(),
                    html.B("Purpose: "),
                    html.Span("See if the gap between income groups is persistent. "),
                    html.Br(),
                    html.B("Example insight: "),
                    html.Span("Low-income areas often show longer resolution times.")
                ], style=insight_bottom_style)
            ], style={**card_style, "marginBottom": "8px"}),

            html.Div([
                html.Div("4. AVERAGE RESOLUTION TIME BY COMPLAINT TYPE AND INCOME GROUP", style=header_style),
                dcc.Graph(id="bar_chart", config={"displayModeBar": False}),
                html.Div([
                    html.B("What it shows: "),
                    html.Span("Resolution time across complaint types by income group. "),
                    html.Br(),
                    html.B("Purpose: "),
                    html.Span("Identify which complaint types show the biggest inequality. "),
                    html.Br(),
                    html.B("Example insight: "),
                    html.Span("Heat/Hot Water shows the largest resolution-time difference.")
                ], style=insight_bottom_style)
            ], style=card_style)

        ], style={
            "flex": "1",
            "padding": "8px",
            "backgroundColor": "white",
            "fontFamily": "Arial"
        })

    ], style={"display": "flex"}),

    html.Div([
        html.B("Overall Story: "),
        html.Span(
            "The map shows where response times are slower or faster. "
            "The scatterplot shows whether income is associated with those differences. "
            "The time-series shows how the pattern changes over time. "
            "The bar chart highlights which complaint types are most affected."
        )
    ], style={
        "backgroundColor": "#f2f4f7",
        "border": "1px solid #bfc7d1",
        "padding": "10px",
        "fontFamily": "Arial",
        "fontSize": "13px"
    })

], style={
    "maxWidth": "1450px",
    "margin": "0 auto",
    "backgroundColor": "white"
})


@app.callback(
    Output("map_chart", "figure"),
    Output("scatter_chart", "figure"),
    Output("time_chart", "figure"),
    Output("bar_chart", "figure"),
    Input("borough_filter", "value"),
    Input("complaint_filter", "value"),
    Input("income_filter", "value"),
    Input("date_filter", "start_date"),
    Input("date_filter", "end_date")
)
def update_dashboard(selected_boroughs, selected_complaints, selected_income, start_date, end_date):
    dff = filter_data(selected_boroughs, selected_complaints, selected_income, start_date, end_date)

    return (
        make_map(dff),
        make_scatter(dff),
        make_time(dff),
        make_bar(dff)
    )


app.run(debug=True)
```
